```python
# W5A16

def armstrong(a):
    b = a
    lst = []
    count = 0
    while b!=0:
        lst.append(b%10)
        count+=1
        b//=10
    for i in lst:
        b+=i**count
    return b == a
armstrong(int(input()))
```

```python
# W5A19

def magic(a):
    sum  = a
    while sum>=10:
        a = sum
        sum = 0
        while a!=0:
            sum+=a%10
            a//=10
    return sum==1
magic(int(input()))
```

```python
# W5A25

def same_half(p1,p2,p3,p4):
    n = (p1[1]-p2[1],p2[0]-p1[0])
    point3 = n[0]*(p3[0]-p1[0])+n[1]*(p3[1]-p1[1])
    point4 = n[0]*(p4[0]-p1[0])+n[1]*(p4[1]-p1[1])
    return point3*point4>=0
def point_in_quadrilateral(quad, p):
    retrn = True
    x = float(quad[0][0]+quad[1][0]+quad[2][0]+quad[3][0])/4
    y = float(quad[0][1]+quad[1][1]+quad[2][1]+quad[3][1])/4
    center = (x,y)
    for i in range(4):
        retrn&=same_half(quad[i],quad[(i+1)%4],center,p)
    return retrn
point_in_quadrilateral([(0,0), (4,0), (4,3),(0,3)],(4,3.1))
```

```python
# W5A28

import math

def angle(p1,p2,p3):
    ba = (p1[0]-p2[0],p1[1]-p2[1])
    bc = (p3[0]-p2[0],p3[1]-p2[1])
    lenBA = (ba[0]**2+ba[1]**2)**0.5
    lenBC = (bc[0]**2+bc[1]**2)**0.5
    cosABC = (ba[0]*bc[0]+ba[1]*bc[1])/lenBA/lenBC
    return math.acos(cosABC)*180/math.pi
angle((1,0),(0,0),(0,1))
```

```python
# W5A29

def list_of_max_min(lst):
    max = lst[0]
    min = lst[0]
    for i in lst:
        if max<i:
            max = i
        if min>i:
            min = i
    return [max,min]
list_of_max_min(map(int,input().split()))
```

```python
# W5A33

def div_ceil(m,n):
    if m%n==0:
        return m//n
    return m//n+1
def smart_taxi_fare(car_type, start_hour, km, wait_minutes):
    dic = {"4 cho":[11_000,13_000,10_500],"7 cho":[12_000,14_000,12_000]}
    specialStartHour = [22,23,0,1,2,3,4,5]
    lookUp = dic[car_type]
    money = lookUp[0]+div_ceil(wait_minutes,5)*2_000
    if km>0.8:
        money += lookUp[1]*min(km-0.8,20)
    if km>20:
        money += lookUp[2]*(km-20)
    if start_hour in specialStartHour:
        money*=1.2
    if km>30 and start_hour not in specialStartHour:
        money*=0.9
    if money<50_000 and km>5:
        return 50_000
    return money
smart_taxi_fare("4 cho",5,0.5,3)
```

```python
# W5A34

def is_unique_num(n):
    sett = set()
    while n!=0:
        char = n%10
        if char in sett:
            return False
        sett.add(char)
        n//=10
    return True
def sum_1_to_n(n):
    sum = 0
    for i in range(1,n+1):
        if is_unique_num(i):
            sum+=i
    return sum
sum_1_to_n(int(input()))
```
