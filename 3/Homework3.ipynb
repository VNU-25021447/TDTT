```python
# W3A1

def reverse(n):
    rev = 0
    while n!=0:
        rev = rev*10+n%10
        n//=10
    return rev

n=int(input())
print(reverse(n))
```

```python
# W3A2
def swap(a,b):
    a^=b
    b^=a
    a^=b
    return a,b

a,b=map(int,input().split())
print(swap(a,b))
```

```python
# W3A5

def div_ceil(m,n):
    if m%n==0:
        return m//n
    return m//n+1

m,n=map(int,input().split())
print(div_ceil(m,n))
```

```python
# W3A9

def can_make_triangle(a,b,c):
    a,b,c = sorted([a,b,c])
    if a+b>c:
        return "Yes"
    return "No"
a,b,c=map(float,input().split())
print(can_make_triangle(a,b,c))
```

```python
# W3A11

def triangle_check(a,b,c):
    a,b,c = sorted([a,b,c])
    if a+b<=c:
        return "Không phải tam giác"
    elif a==b==c:
        return "Tam giác đều" 
    elif a==b or b==c:       
        return "Tam giác cân"
    return "Tam giác thường"
a,b,c=map(int,input().split())
print(triangle_check(a,b,c))
```

```python
# W3A12

def leap_year(n):
    if (n%4==0 and n%100!=0) or n%400==0:
        return "Yes"
    return "No"

n=int(input())
print(leap_year(n))
```
