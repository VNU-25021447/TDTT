```python
# W6A1

def trim_list(lst:list)->list:
    t_lst = []
    for i in lst:
        if i not in t_lst:
            t_lst.append(i)
    return t_lst
lst = input().split()
trim_list(lst)
```

```python
# W6A2

def cummulative_list(lst:list)->list:
    c_list = [lst[0]]
    for i in range(1,len(lst)):
        c_list.append(c_list[i-1]+lst[i])
    return c_list
lst = input().split()
cummulative_list(lst)
```

```python
# W6A3

def left_shift(lst:list,k:int)->list:
    n=k%len(lst)
    return lst[n:]+lst[:n]
lst = [int(i) for i in input().split()]
shift = int(input())
left_shift(lst,shift)
```

```python
# W6A4

def list_dic(lst:list)->dict:
    dic = {}
    for i in lst:
        a,b = i.split(":")
        if a not in dic:
            dic[a] = []
        dic[a]+=[b]
    return dic
lst = input().split()
list_dic(lst)
```

```python
# W6A5

def pos_neg_zero(lst:list)->dict:
    dic = {"positives":0,"negatives":0,"zeros":0}
    for i in lst:
        if i>0:
            dic["positives"]+=1
        elif i<0:
            dic["negatives"]+=1
        else:
            dic["zeros"]+=1
    return dic
lst = [int(i) for i in input().split()]
pos_neg_zero(lst)

```

```python
# W6A9

def merge(dic1:dict,dic2:dict)->dict:
    for i,j in dic2.items():
        if i in dic1:
            dic1[i]+=j
        else:
            dic1[i]=j
    return dic1
def get_dic()->dict:
    dic = {}
    for i in input().split():
        if len(lst:=i.split(":"))==2:
            if not lst[1].isnumeric():
                continue
            dic[lst[0]]=int(lst[1])
    return dic
dic1 = get_dic()
dic2 = get_dic()
merge(dic1,dic2)
```

```python
# W6A10

def combinations(lst:list,k:int)->list:
    c_lst = []
    for i in range(len(lst)):
        if k-lst[i] in lst[i+1:]:
            c_lst.append((lst[i],k-lst[i]))
    return c_lst
lst = [int(i) for i in input().split()]
k = int(input())
combinations(lst,k)
```

```python
# W6A12

def most_occurence(lst:list)->int:
    dic = {}
    max = lst[0]
    prev = 0
    for i in lst:
        if i not in dic:
            dic[i]=0
        dic[i]+=1
        if prev<dic[i]:
            max = i
            prev = dic[i]
    return max
most_occurence([int(i) for i in input().split()])
```

```python
# W6A21

def word_count_dic(string:str)->dict:
    dic = {}
    for i,j in enumerate(string.split()):
        if j not in dic:
            dic[j]=i
        elif isinstance(dic[j],int):
            dic[j]=(dic[j],i)
        else:
            temp = [v for v in dic[j]]
            temp.append(i)
            dic[j]=tuple(temp)
    return dic
word_count_dic(input())

```

```python
# W6A26

def sum_local_max(lst:list)->int:
    sum = 0
    for i in range(1,len(lst)-1):
        if lst[i-1]<lst[i]>lst[i+1]:
            sum+=lst[i]
    return sum
sum_local_max([int(i) for i in input().split()])
```

```python
# W6A28

def true_slope(lst:list)->bool:
    up = lst[0]<lst[1]
    for i in range(1,len(lst)-1):
        if up!=(lst[i]<lst[i+1]):
            return False
    return True
true_slope([int(i) for i in input().split()])
```
