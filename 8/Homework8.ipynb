```python
# W8A16

class Shape():
    def __init__(this):
        pass
    def area():
        pass
class Rectangle(Shape):
    def __init__(this,w,h):
        super().__init__()
        this.w = w
        this.h = h
    def area(this):
        return this.w * this.h
class Circle(Shape):
    def __init__(this,r):
        super().__init__()
        this.r = r
    def area(this):
        return 3.14*this.r**2
match(input()):
    case "RECTANGLE":
        w,h = map(int,input().split())
        r = Rectangle(w,h)
        print(r.area())
    case "CIRCLE":
        r = int(input())
        c = Circle(r)
        print(c.area())
```

```python
# W8A18

class Polygon():
    def perimeter(this):
        pass
class Triangle(Polygon):
    def __init__(this,a,b,c):
        super().__init__()
        this.a = a
        this.b = b
        this.c = c
    def perimeter(this):
        return this.a+this.b+this.c
t = Triangle(*map(int,input().split()))
print(t.perimeter())
```

```python
# W8A19

class Person():
    def __init__(this,name,age):
        this.name = name
        this.age = age
    def describe(this):
        return f"{this.name} {this.age}"
class Student(Person):
    def __init__(this, brand, speed,seats):
        super().__init__(brand, speed)
        this.seats = seats
    def describe(this):
        return f"{this.name} {this.age} {this.school}"
match(input()):
    case "Person":
        name,age = map(str,input().split())
        age = int(age)
        p = Person(name,age)
        print(p.describe())
    case "Student":
        name,age,school = map(str,input().split())
        age = int(age)
        s = Student(name,age,school)
        print(s.describe())
```

```python
# W8A20

class Account():
    def __init__(this,balance):
        this.balance = balance
    def deposit(this,x):
        this.balance+=x
    def get_balance(this):
        return round(this.balance,2)
class SavingAccount(Account):
    def __init__(this,balance,rate):
        super().__init__(balance)
        this.rate = rate
    def add_interest(this):
        this.balance += this.balance * this.rate
s = SavingAccount(*map(float,input().split()))
s.add_interest()
print(s.get_balance())
```

```python
# W8A21

class Printer():
    def print_value(this,x):
        print(x)
class BinaryPrinter(Printer):
    def print_value(this,x):
        print(bin(x))
b = BinaryPrinter()
b.print_value(int(input()))
```

```python
# W8A22

class Matrix2x2():
    def __init__(this,a,b,c,d):
        this.a,this.b,this.c,this.d = a,b,c,d
    def trace(this):
        return this.a+this.d
    def determinant(this):
        return this.a*this.d-this.b-this.c
m = Matrix2x2(*map(int,input().split()))
print(m.trace(),m.determinant())
```

```python
# W8A23

class Polynomial():
    def __init__(this,a,b):
        this.a,this.b = a,b
    def evaluate(this,x):
        return this.a*x+this.b
a,b,x = map(int,input().split())
p = Polynomial(a,b)
print(p.evaluate(x))
```

```python
# W8A24

class Polynomial2():
    def __init__(this,a,b,c):
        this.a,this.b,this.c = a,b,c
    def evaluate(this,x):
        return this.a*x*x+this.b*x+this.c
a,b,c,x = map(int,input().split())
p = Polynomial2(a,b,c)
print(p.evaluate(x))
```

```python
# W8A27

class Product():
    def __init__(this,name,price):
        this.name,this.price = name,price
    def final_price(this):
        return this.price
class DiscountProduct(Product):
    def __init__(this, name, price,percent):
        super().__init__(name, price)
        this.percent = percent
    def final_price(this):
        return round(this.price*(100-this.percent)/100,2)
name,price,percent =  map(str,input().split())
price = int(price)
percent = float(percent)
d = DiscountProduct(name,price,percent)
print(d.final_price())
```

```python
# W8A28

class Counter():
    def __init__(this):
        this.count = 0
    def inc(this):
        this.count+=1
    def dec(this):
        this.count-=1
    def value(this):
        return this.count
c = Counter()
n = int(input())
for i in input().split():
    match i:
        case "INC":
            c.inc()
        case "DEC":
            c.dec()
print(c.value())
```

```python
# W8A29

class Time():
    def __init__(this,h,m,s):
        this.h,this.m,this.s = h,m,s
    def to_seconds(this):
        return this.h*3600+this.m*60+this.s
t = Time(*map(int,input().split()))
print(t.to_seconds())
```
