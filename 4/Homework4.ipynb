```python
# W4A8

def reverse(a):
    rev = 0
    while a!=0:
        rev = rev*10+a%10
        a//=10
    return rev
def palindrome(n):
    count = 0
    while n!=reverse(n):
        count+=1
        n+=reverse(n)
    return count,n
palindrome(int(input()))
```

```python
# W4A9

def is_unique_num(n):
    sett = set()
    while n!=0:
        char = n%10
        if char in sett:
            return False
        sett.add(char)
        n//=10
    return True
def perfect_squares_unique(n):
    lst = []
    for i in range(1,int(n**0.5)+1):
        if is_unique_num(i**2):
            lst.append(i**2)
    return lst
perfect_squares_unique(int(input()))
```

```python
# W4A10

def collatz(n):
    count = 1
    while n!=1:
        count+=1
        if n%2==0:
            n/=2
        else:
            n = 3*n+1
    return count
def longest_collatz(n):
    x = 0
    longest = 0
    for i in range(1,n+1):
        length = collatz(i)
        if longest<length:
            x = i
            longest = length
    return x,longest
longest_collatz(int(input()))
```

```python
# W4A13

def close_pair(a,b):
    sum = 0
    for i in range(1,a):
        if a%i==0:
            sum+=i
    if sum==b:
        sum = 0
        for i in range(1,b):
            if b%i==0:
                sum+=i
        return sum==a
    return False
close_pair(int(input()),int(input()))
```

```python
# W4A14

def biggest_cofactor(m,n):
    max = 0
    for i in range(2,min(m,n)):
        if m%i==0 and n%i==0:
            max = i
    return max
biggest_cofactor(int(input()),int(input()))
```

```python
# W4A26

def fibonaci(n):
    current = 1
    next = 2
    while next<=n:
       next+=current
       current = next-current
    return current
fibonaci(int(input()))
```

```python
# W4A29

def swap_case(string):
    swaped = ""
    for i in string:
        if i.islower():
            temp = i.upper()
        elif i.isupper():
            temp = i.lower()
        else:
            temp = i
        swaped+=temp
    return swaped
swap_case(input())
```

```python
# W4A31

def upper_lower_num_digits(string):
    upper = 0
    lower = 0
    num = 0
    for i in string:
        if i.islower():
            lower+=1
        if i.isupper():
            upper+=1
        if i.isnumeric():
            num+=1
    return upper,lower,num
upper_lower_num_digits(input())
```
