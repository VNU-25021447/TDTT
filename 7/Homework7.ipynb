```python
# W7A1

def binary_search(arr,left,right,target):
    if left>right:
        return -1
    mid = (left + right)//2
    if arr[mid] > target:
        return binary_search(arr,left,mid-1,target)
    elif arr[mid] < target:
        return binary_search(arr,mid+1,right,target)
    return mid
arr = [1,3,4,6,7,9,100,1000]
x = 6
print(binary_search(arr,0,len(arr)-1,x))
```

```python
# W7A3

def selection_sort(lst):
    l = len(lst)
    for i in range(0,l):
        min = i
        for j in range(i+1,l):
            if lst[min]>lst[j]:
                min = j
        lst[i],lst[min] = lst[min],lst[i]
    return lst

lst = [1,3,5,4,2,8]
print(selection_sort(lst))
```

```python
# W7A7

def increment_length(lst):
    max = 0
    for i in range(0,len(lst)):
        l = 1
        prev = lst[i]
        for j in lst[i:]:
            if prev<j:
                prev = j
                l+=1
        if max<l:
            max = l
    return max

print(increment_length([7,7,7,7,7,7,7,7]))
```

```python
# W7A8

def query(i,q):
    retrn = []
    for j in q:
        lst = []
        for k in i:
            if k[0]<=j<=k[1]:
                lst.append(k[1]-k[0]+1)
        lst.sort()
        retrn.append(-1 if len(lst)==0 else lst[0])
    return retrn

print(query([[2, 8], [1, 5], [10, 15]],[1, 4, 6, 11, 20]))
```

```python
# W7A10

def summodulo(lst,num):
    max = 0
    for i in range(len(lst)):
        sum = 0
        for j in lst[i:]:
            sum+=j
            if max<sum%num:
                max = sum%num
    return max
print(summodulo([7,11,14,2,8,4],12))
```

```python
# W7A13

def harmonic_list(lst):
    dic = {}
    for i in range(len(lst)):
        dic[lst[i]] = dic[lst[i]]+[i] if lst[i] in dic else [i]
    max = 0
    key = 0
    for i in dic.keys():
        try:
            k = len(dic[i])+len(dic[i+1])
        except KeyError:
            continue
        if k>max:
            max = k
            key = i
    o = []
    lst2 = dic[key]
    lst3 = dic[key+1]
    for i in range(max):
        if len(lst2)==0:
            o.append(key+1)
        elif len(lst3)==0:
            o.append(key)
        elif lst2[0]<lst3[0]:
            o.append(key)
            lst2.pop(0)
        elif lst2[0]>lst3[0]:
            o.append(key+1)
            lst3.pop(0)
    return o

lst = [int(i) for i in input().split()]
print(harmonic_list(lst))
```

```python
# W7A18

def add_sorted(lst,target,left = 0,right = len(lst)-1):
    if left>=right:
        lst.insert(left+1 if left==right else left,target)
        return lst
    mid = (left + right)//2
    if lst[mid] < target:
        if target < lst[mid+1]:
            lst.insert(mid+1,target)
            return lst
        else:
            return add_sorted(lst,target,mid+1,right)
    else:
        return add_sorted(lst,target,left,mid-1)
def conditional_sort(lst):
    eve = []
    odd = []
    for i in lst:
        if i%2==0:
            add_sorted(eve,i,0,len(eve)-1)
        else:
            add_sorted(odd,i,0,len(odd)-1)
    return eve+[i for i in reversed(odd)]
print(conditional_sort([int(i) for i in input().split()]))
```

```python
# W7A19

def merge_list(lst1:list,lst2:list):
    lst = []
    increase = lst1[0]<lst1[1]
    while len(lst1)!=0 or len(lst2)!=0:
        if len(lst1)==0:
            lst.append(lst2[0])
            lst2.pop(0)
        elif len(lst2)==0:
            lst.append(lst1[0])
            lst1.pop(0)
        elif lst1[0]<lst2[0]:
            lst.append(lst1[0] if increase else lst2[0])
            lst1.pop(0) if increase else lst2.pop(0)
        elif lst1[0]>lst2[0]:
            lst.append(lst2[0] if increase else lst1[0])
            lst2.pop(0) if increase else lst1.pop(0)
        else:
            lst.append(lst1[0])
            lst.append(lst2[0])
            lst1.pop(0)
            lst2.pop(0)
    return lst
print(merge_list([int(i) for i in input().split()],[int(i) for i in input().split()]))
```

```python
# W7A27

def valid_list(s):
    lst = []
    dic = {"(":")","[":"]","{":"}","<":">",}
    for i in s:
        if i in dic:
            lst.append(i)
            continue
        try:
            if dic[lst[len(lst)-1]]==i:
                lst.pop(len(lst)-1)
        except IndexError:
            continue
    return "YES" if len(lst)==0 else "NO"
print(valid_list(input()))
```

```python
# W7A28

def is_prime(num):
    for i in range(2,num//2):
        if num%i==0:
            return False
    return True
def longest(n):
    lst = []
    for i in range(n):
        if not is_prime(i):
            continue
        temp = []
        for j in range(i+1,n):
            if is_prime(j):
                lst.append(temp)
                temp = []
                break
            else:
                temp.append(j)
        else:
            lst.append(temp)
    max = []
    for i in lst:
        if len(max)<len(i):
            max = i
    return max
print(longest(int(input())))
```

```python
# W7A29

def increasing_list(lst):
    lst1 = []
    prev = lst[0]
    for i in range(len(lst)):
        temp = []
        prev = lst[i]-1
        for j in lst[i:]:
            if prev<j:
                temp.append(j)
                prev = j
            else:
                break
        lst1.append(temp)
    max = []
    for i in lst1:
        if len(max)<len(i):
            max = i
    return max

print(increasing_list([int(i) for i in input().split()]))
```

```python
# W7A31

def swap(lst,num,pos1,pos2):
    l = len(lst)-1
    sw = lst[l-pos1]-lst[l-pos2]
    return num + sw*(10**pos2-10**pos1)
def min_swap(s):
    lst = []
    temp = s
    while temp>0:
        lst.insert(0,temp%10)
        temp//=10
    min = swap(lst,s,0,1)
    l = len(lst)
    complexity = 0
    for i in range(l):
        for j in range(i+1,l):
            num = swap(lst,s,i,j)
            if min>num:
                min = num
            complexity+=1

    return min,complexity
print(min_swap(1232329899999999991219999999))
```
